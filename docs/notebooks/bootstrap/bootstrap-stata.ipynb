```stata
local root : environment FINECO_BOOTSTRAP_ROOT
cd "`root'"
capture log close _all
log using "logs/bootstrap/v10/stata.log", text replace
set more off
display "Stata version: " c(stata_version)
sysuse "auto.dta", clear
summarize price
mean price
regress price weight mpg, noheader
```

```stata
do "tools/bootstrap/01-tiny-bootstrap.do"
```

```stata
clear
input double x
2
4
5
7
12
end
capture program drop boot_mean_once
program define boot_mean_once, rclass
    preserve
    bsample
    quietly summarize x
    return scalar xbar = r(mean)
    restore
end
simulate xbar_star=r(xbar), reps(9999) seed(12345): boot_mean_once
summarize xbar_star, detail
centile xbar_star, centile(2.5 97.5)
save "logs/bootstrap/v10/tiny-stata-replicates.dta", replace
```

```stata
sysuse "auto.dta", clear
regress price weight length, ///
    vce(bootstrap, reps(5000) seed(13579))
scalar vce_se = _se[weight]
bootstrap beta_weight=_b[weight], ///
    reps(5000) seed(13579): ///
    regress price weight length
assert abs(_se[beta_weight] - vce_se) < 1e-8
```
